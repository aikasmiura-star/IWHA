```python
# Welcome to your new notebook
# Type here in the cell editor to add code!
from pyspark.sql import functions as F

TABLE = "emodnet_ml_scored"
OUT_DISCORD = "emodnet_ml_discordance"

df = spark.table(TABLE)

df_discord = df.filter(
    F.col("Pred_BaggedTrees") != F.col("Pred_WeightedKNN")
)

total = df.count()
discord = df_discord.count()

print(f"Total registros: {total}")
print(f"Discordâncias: {discord}")
print(f"Taxa de discordância: {discord/total:.4%}")

# Salvar
(
    df_discord.write
      .mode("overwrite")
      .format("delta")
      .saveAsTable(OUT_DISCORD)
)

print(f"✅ Tabela criada: {OUT_DISCORD}")
```
